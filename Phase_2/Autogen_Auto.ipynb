```python
import autogen
# Required imports for LangChain
import warnings
# Suppress only DeprecationWarnings
warnings.filterwarnings("ignore")
```

```python
llm_config = {"model": "gpt-3.5-turbo"}
```

```python
# ANSI escape code for green text
GREEN = "\033[92;1m"
BLUE_BOLD = "\033[94;1m"
RESET = "\033[0m"
```

```python
customer_statement="""
I want to create an educational portal where students and teachers can interact.
The portal should include features such as course management, assignment submission, grading system, 
and discussion forums. Teachers should be able to create and manage courses, while students should be able
to enroll, submit assignments, and receive feedback. The system should also support document sharing and 
administrative controls. 
"""
```

```python
# Define all agents
customer_proxy = autogen.ConversableAgent(
    name="Customer",
    system_message="""You are the Customer. Clearly express your vision and requirements for an educational portal where students and teachers interact. Your input drives the Product Owner's backlog creation. Provide key features and desired outcomes.""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)

product_owner = autogen.ConversableAgent(
    name="Product_Owner",
    system_message="""You are the Product Owner. Break the customer's vision and roadmap. Coordinate with all other agents. After collecting their inputs, generate a final project plan with:
- Total sprints
- User stories with effort (Story Points)
- Sprint-wise assignments
- Total duration in days which excludes weekends (assume 2 weeks per sprint).""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)

scrum_master = autogen.ConversableAgent(
    name="Scrum_Master",
    system_message="""You are the Scrum Master. Define sprint length, max story points per sprint, daily stand-ups, and retrospectives.""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)

ui_ux_designer = autogen.ConversableAgent(
    name="UI_UX_Designer",
    system_message="""Design user interfaces and user experiences based on stories and customer feedback. Provide UI/UX designs for user stories.Provide sprint wise design in following format: Sprint 1, Sprint 2,etc. Align with Architect and Developer.""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)

system_architect = autogen.ConversableAgent(
    name="System_Architect",
    system_message="""You are the System Architect. Define system architecture, database schema, APIs, and integration flow.Define architecture design required for the project and give sprint wise architecture design in following format: Sprint 1, Sprint 2, etc. Estimate design effort: Effort = Design Pages / Productivity.""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)
software_developer = autogen.ConversableAgent(
    name="Software_Developer",
    system_message="""You are the Software Developer. Develop software as per acceptance criteria and user stories. Estimate number of lines of source code based on user stories(for every sprint). Provide sprint wise plan in following format: Sprint 1, Sprint 2,etc. Use: Effort = SLOC / Productivity.
Collaborate with Architect and UI/UX Designer.""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)

test_engineer = autogen.ConversableAgent(
    name="Test_Engineer",
    system_message="""You are the Test Engineer. Create and execute a test plan covering unit, integration, UI, and UAT tests. 
Estimate testing effort based on user stories and ensure the product meets all acceptance criteria and is defect-free. Provide sprintwise test plan, estimates and testcases in following format: Sprint 1, Sprint 2,etc.Estimate testing effort: Effort = Test Cases / Productivity. Base this on features from Dev and Design.""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)

document_engineer = autogen.ConversableAgent(
    name="Document_Engineer",
    system_message="""You are the Document Engineer. Creates end user documentation, release notes and internal process documentation. 
    Create documentation for user stories for every sprint. Provide sprint wise documentation in the following format: Sprint1, Spprint 2, etc. Estimate documentation effort: Effort = Pages / Productivity. Use inputs from all other agents.""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)
```

```python
groupchat = autogen.GroupChat(
    agents=[
        customer_proxy,
        scrum_master,
        software_developer,
        ui_ux_designer,
        test_engineer,
        system_architect,
        document_engineer,
        product_owner,
        
    ],
    messages=[],
    speaker_selection_method="auto",  
    max_round=10,
)

manager = autogen.GroupChatManager(groupchat=groupchat, llm_config=llm_config)
```

```python
chat_result = product_owner.initiate_chat(
    manager,
    message=customer_statement,
)
```
