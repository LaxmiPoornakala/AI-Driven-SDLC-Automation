```python
# Cell 2: Customer Proxy Agent
customer_agent_prompt = """You are the Customer. Clearly express your vision and requirements for an educational portal 
where students and teachers interact. Your input drives the Product Owner's backlog creation. Provide key 
features and desired outcomes.

Customer's statement: {customer_statement}"""

customer_proxy = LLMChain(
    llm=llm_config,
    prompt=PromptTemplate(
        input_variables=["customer_statement"],
        template=customer_agent_prompt
    )
)
```

```python
# Cell 3: Product Owner Agent
product_owner_prompt = """You are the Product Owner. Break this vision into epics/user stories:
{customer_statement}
Generate final plan with:
- Total sprints (2 weeks/sprint)
- User stories with Story Points
- Sprint assignments
- Total duration (days)
Include explicit calculations."""

product_owner = LLMChain(
    llm=llm_config,
    prompt=PromptTemplate(
        input_variables=["customer_statement"],
        template=product_owner_prompt
    )
)
```

```python
# Cell 4: Scrum Master Agent
scrum_master_prompt = """As Scrum Master, define for {customer_statement}:
- Sprint length
- Max story points/sprint
- Daily stand-up structure
- Retrospective format
Ensure team velocity considered."""

scrum_master = LLMChain(
    llm=llm_config,
    prompt=PromptTemplate(
        input_variables=["customer_statement"],
        template=scrum_master_prompt
    )
)
```

```python
# Cell 5: Software Developer Agent
developer_prompt = """As Developer for {customer_statement}:
- Break user stories into technical tasks
- Calculate effort: SLOC/Productivity
- Implement frontend/backend/database
Collaborate with UI/UX and Architect."""

software_developer = LLMChain(
    llm=llm_config,
    prompt=PromptTemplate(
        input_variables=["customer_statement"],
        template=developer_prompt
    )
)
```

```python
# Cell 6: UI/UX Designer Agent
ui_ux_prompt = """As UI/UX Designer for {customer_statement}:
- Design accessible/responsive interfaces
- Create wireframes
- Estimate effort: UI Screens/Productivity
Align with Developer/Architect."""

ui_ux_designer = LLMChain(
    llm=llm_config,
    prompt=PromptTemplate(
        input_variables=["customer_statement"],
        template=ui_ux_prompt
    )
)
```

```python
# Cell 7: Test Engineer Agent
tester_prompt = """As Test Engineer for {customer_statement}:
- Create test plan (unit/integration/UI/UAT)
- Estimate effort: Test Cases/Productivity
- Use inputs from Dev/Design"""

test_engineer = LLMChain(
    llm=llm_config,
    prompt=PromptTemplate(
        input_variables=["customer_statement"],
        template=tester_prompt
    )
)
```

```python
# Cell 8: System Architect Agent
architect_prompt = """As System Architect for {customer_statement}:
- Define architecture/database schema/APIs
- Create integration flow
- Estimate effort: Design Pages/Productivity"""

system_architect = LLMChain(
    llm=llm_config,
    prompt=PromptTemplate(
        input_variables=["customer_statement"],
        template=architect_prompt
    )
)
```

```python
# Cell 9: Document Engineer Agent
doc_engineer_prompt = """As Document Engineer for {customer_statement}:
- Prepare user/admin/API docs
- Estimate effort: Pages/Productivity
- Use inputs from all agents"""

document_engineer = LLMChain(
    llm=llm_config,
    prompt=PromptTemplate(
        input_variables=["customer_statement"],
        template=doc_engineer_prompt
    )
)
```

```python
# Cell 10: Execution Workflow
customer_statement = """
I want to create an educational portal where students and teachers can interact.
The portal should include features such as course management, assignment submission, grading system, 
and discussion forums. Teachers should be able to create and manage courses, while students should be able
to enroll, submit assignments, and receive feedback. The system should also support document sharing and 
administrative controls.
"""

agents = [
    ("Customer", customer_proxy),
    ("Scrum_Master", scrum_master),
    ("Software_Developer", software_developer),
    ("UI_UX_Designer", ui_ux_designer),
    ("Test_Engineer", test_engineer),
    ("System_Architect", system_architect),
    ("Document_Engineer", document_engineer),
    ("Product_Owner", product_owner)
]

def run_group_chat(agents, customer_statement, max_round=8):
    conversation = []
    for round_num in range(max_round):
        for role, agent in agents:
            response = agent.run(customer_statement=customer_statement)
            conversation.append(f"{role} (to chat_manager):\n\n{response}\n\n--------------------------------------------------------------------------------")
    return conversation

# Run conversation
conversation = run_group_chat(agents, customer_statement)

# Print results
for msg in conversation:
    print(msg)
    print("\n" + "="*80 + "\n")
```
