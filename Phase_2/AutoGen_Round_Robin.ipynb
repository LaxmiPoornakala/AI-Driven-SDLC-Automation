```python
import autogen
```

```python
llm_config={"model": "gpt-4o-mini"}
```

```python
customer_statement="""
I want to create an educational portal where students and teachers can interact.
The portal should include features such as course management, assignment submission, grading system, 
and discussion forums. Teachers should be able to create and manage courses, while students should be able
to enroll, submit assignments, and receive feedback. The system should also support document sharing and 
administrative controls.
"""
```

```python
customer_proxy = autogen.ConversableAgent(
    name="Customer",
    system_message="""You are the Customer. Clearly express your vision and requirements for an educational portal "
    "where students and teachers interact. Your input drives the Product Owner's backlog creation. Provide key " 
    "features and desired outcomes.""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)
```

```python
product_owner = autogen.ConversableAgent(
    name="Product_Owner",
    system_message="""You are the Product Owner. Break the customer vision into epics and user stories. 
Coordinate with all other agents. After collecting their inputs, generate a final project plan with Total sprints, User stories 
with effort (Story Points), Sprint-wise assignments and calculate total duration in days (assume 2 weeks per sprint and working hours per week) excluding weekends.""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)

scrum_master = autogen.ConversableAgent(
    name="Scrum_Master",
    system_message="""You are the Scrum Master. Define Agile processes: sprint length, max story points per sprint, daily stand-ups, and retrospectives. 
Ensure coordination and team velocity are considered in planning.""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)

software_developer = autogen.ConversableAgent(
    name="Software_Developer",
    system_message="""You are the Software Developer. Translate user stories into technical tasks and estimate development effort.
Implement frontend, backend, and database logic. Use: Effort = SLOC / Productivity.
Collaborate with Architect and UI/UX Designer.""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)

ui_ux_designer = autogen.ConversableAgent(
    name="UI_UX_Designer",
    system_message="""You are the UI/UX Designer. Design accessible, responsive user interfaces for students and teachers.
Create screen flows and wireframes. Estimate UI effort per screen. Align with Architect and Developer.""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)

test_engineer = autogen.ConversableAgent(
    name="Test_Engineer",
    system_message="""You are the Test Engineer. Create a test plan with unit, integration, UI, and UAT tests.
Estimate testing effort: Effort = Test Cases / Productivity. Base this on features from Dev and Design.""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)

system_architect = autogen.ConversableAgent(
    name="System_Architect",
    system_message="""You are the System Architect. Define system architecture, database schema, APIs, and integration flow.
Estimate design effort: Effort = Design Pages / Productivity. Provide diagrams and technical blueprint.""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)

document_engineer = autogen.ConversableAgent(
    name="Document_Engineer",
    system_message="""You are the Document Engineer. Prepare user manuals, admin guides, API docs, and onboarding material.
Estimate documentation effort: Effort = Pages / Productivity. Use inputs from all other agents.""",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="NEVER",
)
```

```python
# Initialize GroupChat
groupchat = autogen.GroupChat(
    agents=[
        customer_proxy,
        scrum_master,
        software_developer,
        ui_ux_designer,
        test_engineer,
        system_architect,
        document_engineer,
        product_owner,
    ],
    messages=[],
    speaker_selection_method="round_robin",  # Automatically selects the next speaker
    max_round=8,
)
```

```python
manager = autogen.GroupChatManager(
    groupchat=groupchat, 
    llm_config=llm_config
)
```

```python
chat_result = customer_proxy.initiate_chat(
    manager,
    message=customer_statement,
)
```
